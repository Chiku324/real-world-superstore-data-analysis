This notebook follows a complete analyst workflow: data inspection → cleaning → EDA → KPI calculation → visualization → recommendations.

```
import os, pandas as pd, numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
DATA_PATH = "../data/superstore.csv"
df = pd.read_csv(DATA_PATH)
df.head()
```

```
print("Rows, columns:", df.shape)
print("\nMissing values:")
print(df.isna().sum())
print("\nDuplicate rows:", df.duplicated().sum())
print("\nData types:")
print(df.dtypes)
```

```
df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], errors="coerce")
df["Profit Margin"] = np.where(df["Sales"] != 0, df["Profit"]/df["Sales"], np.nan)
df["Order Year"] = df["Order Date"].dt.year
df["Order Month"] = df["Order Date"].dt.to_period("M").astype(str)
kpis = {
    "Sales": df["Sales"].sum(),
    "Profit": df["Profit"].sum(),
    "Profit Margin": df["Profit"].sum()/df["Sales"].sum(),
    "Quantity": df["Quantity"].sum(),
    "Orders": df["Order ID"].nunique(),
    "Customers": df["Customer ID"].nunique(),
    "Products": df["Product ID"].nunique()
}
kpis
```

```
fig, ax = plt.subplots(figsize=(10,5))
df.groupby("Order Month")["Sales"].sum().plot(ax=ax)
ax.set_title("Monthly Sales Trend")
ax.set_ylabel("Sales ($)")
plt.xticks(rotation=45)
plt.tight_layout()
```

```
fig, ax = plt.subplots(figsize=(8,5))
df.groupby("Category")["Sales"].sum().sort_values(ascending=False).plot(kind="bar", ax=ax)
ax.set_title("Sales by Category")
ax.set_ylabel("Sales ($)")
plt.xticks(rotation=0)
plt.tight_layout()
```

```
fig, ax = plt.subplots(figsize=(8,5))
df.groupby("Region")["Sales"].sum().sort_values(ascending=False).plot(kind="bar", ax=ax)
ax.set_title("Sales by Region")
ax.set_ylabel("Sales ($)")
plt.xticks(rotation=0)
plt.tight_layout()
```

```
profit_by_sub = df.groupby("Sub-Category")["Profit"].sum().sort_values()
profit_by_sub.plot(kind="barh", figsize=(9,7), title="Profit by Sub-Category")
plt.xlabel("Profit ($)")
plt.tight_layout()
```

```
sample = df.sample(min(3000, len(df)), random_state=42)
sns.scatterplot(data=sample, x="Discount", y="Profit", hue="Category", alpha=0.7)
plt.title("Discount vs Profit")
plt.tight_layout()
```
